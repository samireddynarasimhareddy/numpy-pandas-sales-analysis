# 02 — Pandas Fundamentals

Load the supplied synthetic sales CSV and practise inspection, selection, filtering, sorting, and aggregation.

```python
from pathlib import Path
import pandas as pd

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
df = pd.read_csv(ROOT / 'data' / 'raw' / 'sales.csv')
df.head()
```

```python
print('Shape:', df.shape)
print('Columns:', df.columns.tolist())
print(df.dtypes)
print(df.isna().sum())
df.describe(include='all')
```

## Selection and filtering

```python
print(df[['Product', 'Quantity', 'Unit_Price']].head())
print(df.loc[0, 'Product'])
print(df.iloc[0, 0])
high_qty = df.loc[pd.to_numeric(df['Quantity'], errors='coerce') >= 4]
print(high_qty.head())
```

## Sorting and frequency counts

```python
print(df['Product'].value_counts(dropna=False))
print(df.sort_values('Order_Date').head())
print(df['Region'].nunique(dropna=False))
```

## Grouping and aggregation

```python
work = df.copy()
work['Quantity'] = pd.to_numeric(work['Quantity'], errors='coerce')
work['Unit_Price'] = pd.to_numeric(work['Unit_Price'], errors='coerce')
work['Gross_Sales'] = work['Quantity'] * work['Unit_Price']
work.groupby('Product').agg(
    rows=('Order_ID', 'size'),
    median_quantity=('Quantity', 'median'),
    average_price=('Unit_Price', 'mean'),
    gross_sales=('Gross_Sales', 'sum')
).sort_values('gross_sales', ascending=False)
```
