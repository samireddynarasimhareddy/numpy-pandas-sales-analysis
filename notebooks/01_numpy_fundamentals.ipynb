# 01 — NumPy Fundamentals

Practice arrays, shapes, indexing, vectorization, broadcasting, statistics, random sampling, and outlier detection. Run cells from top to bottom.

```python
import numpy as np

rng = np.random.default_rng(42)
prices = np.array([25.0, 18.0, 45.0, 22.0, 180.0, 65.0])
quantities = np.array([2, 4, 1, 3, 1, 2])
print('Shape:', prices.shape, 'Dimensions:', prices.ndim, 'Size:', prices.size)
print('Revenue by product:', prices * quantities)
print('Total revenue:', np.sum(prices * quantities))
```

## Exercise 2 — Descriptive statistics
Calculate mean, median, standard deviation, variance, quartiles, minimum, maximum, and the index of the highest value. Explain why mean and median can differ.

```python
sales = np.array([50, 60, 70, 80, 90, 500], dtype=float)
print('mean:', np.mean(sales))
print('median:', np.median(sales))
print('std:', np.std(sales))
print('quartiles:', np.percentile(sales, [25, 50, 75]))
print('maximum index:', np.argmax(sales))
```

## Exercise 3 — Boolean masks and `where`
Find sales greater than 75, replace values below 60 with 60 using `np.where`, and count how many values meet a condition.

```python
print(sales[sales > 75])
print(np.where(sales < 60, 60, sales))
print('Count > 75:', np.count_nonzero(sales > 75))
```

## Exercise 4 — IQR outliers
Use the 25th and 75th percentiles to calculate the interquartile range (IQR), then flag values outside Q1 − 1.5×IQR and Q3 + 1.5×IQR. An outlier is not necessarily an error.

```python
q1, q3 = np.percentile(sales, [25, 75])
iqr = q3 - q1
mask = (sales < q1 - 1.5 * iqr) | (sales > q3 + 1.5 * iqr)
print('Q1, Q3, IQR:', q1, q3, iqr)
print('Potential outliers:', sales[mask])
```
